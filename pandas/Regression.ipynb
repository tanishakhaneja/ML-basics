```python
df3 = pd.read_csv('data/hiring.csv')
df3
```

```python
df3['experience'] = df3['experience'].fillna('zero')
df3
```

```python
median_testscore = math.floor(df3['test_score(out of 10)'].median())
median_testscore
```

```python
df3['test_score(out of 10)'] = df3['test_score(out of 10)'].fillna(median_testscore)
df3
```

```python
reg = linear_model.LinearRegression()
reg.fit(df3[['experience','test_score(out of 10)','interview_score(out of 10)']],df3['salary($)'])
```

```python
reg.coef_
```
